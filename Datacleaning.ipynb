```python
 # Data Cleaning Project

#Goal: Clean a messy dataset so it is ready for analysis.
#Dataset: (Customer call list)
#Tools: Python, Pandas
```

```python
# Load the dataset into a table so we can inspect and clean it
import pandas as pd

df = pd.read_excel(r"C:\Users\Lenovo\OneDrive\Documents\Jupiter Python\pandas\Customer Call List.xlsx Data cleaning.xlsx")
```

```python
df
```

```python
# Remove duplicates values
df = df.drop_duplicates()
df
```

```python
# drop any column 
df = df.drop(columns ='Not_Useful_Column')
df
```

```python
# Clean last names: remove stray digits and symbols at the ends
df['Last_Name']= df['Last_Name'].str.strip('123._/')
df
```

```python
# Remove everything that isn't a digit (spaces, dashes, brackets), so every phone number starts in the same plain format
# And  Format the 10 digits as XXX-XXX-XXXX 

df['Phone_Number']= df['Phone_Number'].astype(str).str.replace(r'\D', '', regex = True)

df['Phone_Number'] = df['Phone_Number'].str.replace(r'(\d{3})(\d{3})(\d{4})' , r'\1-\2-\3', regex= True)
```

```python
df
```

```python
# Replace any missing phone numbers with an empty text, so the column has no blank (NaN) values left
df['Phone_Number'] = df['Phone_Number'].fillna('')
df
```

```python
# Split the Address column into three separate columns (Street, State, Zip code), so each piece can be filtered and grouped on its own
df[['Street','State', 'zip_code']]=df['Address'].str.split(',', n=2, expand = True)
df
```

```python
df
```

```python
# Shorten "Yes" to "Y" in the Paying Customer column, so it uses the same short code as the other entries
# Shorten "No" to "N" for the same reason

df['Paying Customer']= df['Paying Customer'].str.replace('Yes', 'Y')
df['Paying Customer']= df['Paying Customer'].str.replace('No', 'N')
df
```

```python
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('Yes', 'Y')
# Shorten "Yes" to "Y" in the Do Not Contact column
```

```python
df
```

```python
df = df.fillna('')
# Replace every missing value (NaN) in the whole table with empty text, so no blank cells cause problems later
```

```python
#Replace the text "N/a" with empty text, so placeholders meaning "no data" are treated the same as other missing values
df = df.replace('N/a', '')
```

```python
df
```

```python
# Remove customers who asked not to be contacted (Do_Not_Contact = Y), so they can't be included in any outreach list
for x in df.index:
    if df.loc[x, 'Do_Not_Contact'] == 'Y':
        df.drop(x, inplace = True)
```

```python
df
```

```python
# Remove customers with no phone number, since they can't be reached by phone and would leave gaps in a call list
for x in df.index:
   if df.loc[x , 'Phone_Number'] == '':
    df.drop(x, inplace = True )
```

```python
df
```

```python
# Shorten "No" to "N" in the Do_Not_Contact column, so it uses the same Y/N code as the rest of the column
df['Do_Not_Contact']= df['Do_Not_Contact'].replace('No', 'N')
```

```python
df
```

```python
# Use CustomerID as the row label, so each customer can be looked up by their unique ID
df = df.set_index('CustomerID')
df
```

```python
df.to_csv('customercall.csv', index=False)
```

```python
## Summary

 #Cleaned names, phone numbers and addresses so each follows one consistent format
 #Standardised Yes/No columns to Y/N
 #Replaced missing values and "N/a" placeholders with empty values
 #Removed customers who asked not to be contacted and those with no phone number
 #Result: a clean table, with CustomerID as the index, ready for analysis or outreach
```
